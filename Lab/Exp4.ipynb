```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy

# Actual weather distribution for 30 days
true_p = np.array([15, 10, 5]) / 30

# Predicted probabilities from two models
model_a = np.array([0.50, 0.30, 0.20])
model_b = np.array([0.70, 0.20, 0.10])

names = ["Sunny", "Cloudy", "Rainy"]

# 1. Calculate entropy
H = entropy(true_p, base=2)

print("Entropy:", H)

# 2. Calculate cross entropy and KL divergence
for name, q in [("Model A", model_a),
                ("Model B", model_b)]:

    cross_entropy = -np.sum(true_p * np.log2(q))

    kl = entropy(true_p, q, base=2)

    print("\n", name)
    print("Cross Entropy:", cross_entropy)
    print("KL Divergence:", kl)

    print("Entropy + KL:", H + kl)

# 3. Verify KL divergence is asymmetric
kl_pq = entropy(true_p, model_a, base=2)
kl_qp = entropy(model_a, true_p, base=2)

print("\nKL(P || Q):", kl_pq)
print("KL(Q || P):", kl_qp)

# 4. Binary Cross Entropy Loss
p = np.linspace(0.001, 0.999, 300)

y_true = 1

loss = -(y_true * np.log(p) +
         (1-y_true) * np.log(1-p))

plt.plot(p, loss)

plt.xlabel("Predicted Probability")
plt.ylabel("Binary Cross Entropy Loss")
plt.title("Binary Cross Entropy Loss")

plt.grid()
plt.show()
```
