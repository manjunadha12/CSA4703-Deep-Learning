```python

import numpy as np
import matplotlib.pyplot as plt
customer_genre = np.array([
    [5, 4, 0, 1, 2],
    [3, 5, 2, 0, 1],
    [0, 1, 5, 4, 3],
    [1, 0, 4, 5, 4],
    [4, 3, 1, 2, 0],
    [2, 2, 3, 1, 5]
], dtype=float)
genre_book = np.array([
    [0.9, 0.1, 0.2, 0.0],
    [0.1, 0.8, 0.1, 0.2],
    [0.0, 0.2, 0.9, 0.1],
    [0.2, 0.0, 0.1, 0.9],
    [0.3, 0.1, 0.2, 0.8]
], dtype=float)

print("CUSTOMER-GENRE MATRIX:")
print(customer_genre)

print("\nGENRE-BOOK MATRIX:")
print(genre_book)
dot = np.dot(customer_genre[0], customer_genre[1])

print("\nDOT PRODUCT OF CUSTOMER 1 AND 2:")
print(dot)

score_matrix = np.matmul(customer_genre, genre_book)

print("\nCUSTOMER-BOOK SCORE MATRIX:")
print(score_matrix)

transpose = customer_genre.T

print("\nTRANSPOSE OF CUSTOMER MATRIX:")
print(transpose)
square_matrix = np.matmul(customer_genre.T, customer_genre)

inverse = np.linalg.inv(square_matrix)

print("\nSQUARE MATRIX:")
print(square_matrix)

print("\nINVERSE MATRIX:")
print(inverse)
addition = customer_genre + customer_genre
subtraction = customer_genre - customer_genre
multiplication = customer_genre * customer_genre

print("\nELEMENT-WISE ADDITION:")
print(addition)

print("\nELEMENT-WISE SUBTRACTION:")
print(subtraction)

print("\nELEMENT-WISE MULTIPLICATION:")
print(multiplication)
norms = np.linalg.norm(
    customer_genre, axis=1, keepdims=True
)

normalized = customer_genre / norms

cosine_similarity = np.matmul(
    normalized, normalized.T
)

print("\nCOSINE SIMILARITY MATRIX:")
print(np.round(cosine_similarity, 3))

# Find the most similar customer pair
upper = np.triu(cosine_similarity, k=1)

i, j = np.unravel_index(
    np.argmax(upper), upper.shape
)

print("\nMOST SIMILAR CUSTOMERS:")
print("Customer", i + 1, "and Customer", j + 1)

print("Similarity Score:", round(cosine_similarity[i, j], 4))

# ---------------------------------------
# 9. DENSE LAYER: y = Wx + b
# ---------------------------------------

# Input vector: Customer 1
x = customer_genre[0]

# Weight matrix
W = genre_book.T

# Bias vector
b = np.array([0.5, 0.5, 0.5, 0.5])

# Dense layer calculation
y = np.dot(W, x) + b

print("\nDENSE LAYER OUTPUT:")
print("Input x:", x)
print("Weight W:\n", W)
print("Bias b:", b)
print("Output y:", y)

# ---------------------------------------
# 10. VISUALIZE SCORE MATRIX
# ---------------------------------------

plt.figure(figsize=(8, 5))

plt.imshow(score_matrix, cmap="YlGnBu", aspect="auto")

plt.colorbar(label="Recommendation Score")

plt.xticks(
    range(4),
    ["Book 1", "Book 2", "Book 3", "Book 4"]
)

plt.yticks(
    range(6),
    ["Customer 1", "Customer 2", "Customer 3",
     "Customer 4", "Customer 5", "Customer 6"]
)

plt.title("Customer-Book Recommendation Score Heatmap")
plt.xlabel("Books")
plt.ylabel("Customers")

# Display values inside heatmap
for i in range(6):
    for j in range(4):
        plt.text(
            j, i,
            round(score_matrix[i, j], 2),
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```
